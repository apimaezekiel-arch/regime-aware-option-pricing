**SYNTHETIC illustration only.** This notebook uses hypothetical contract parameters; no market observations or empirical conclusions.

```python
from pathlib import Path
import sys
root = Path.cwd()
if root.name == "notebooks": root = root.parent
sys.path.insert(0, str(root))
from src.pricing import black_scholes, merton_jump, heston_mc
import pandas as pd
import matplotlib.pyplot as plt
```

## Compare European call valuation at hypothetical parameters
Heston Monte Carlo has simulation error; these models need observed bid/ask quote data for actual comparison.

```python
S, r, T, sigma = 100, .04, .5, .20
strikes = [85, 90, 95, 100, 105, 110, 115]
rows=[]
for K in strikes:
    b=black_scholes(S,K,T,r,sigma)
    m=merton_jump(S,K,T,r,sigma,.8,-.08,.18)
    h,se=heston_mc(S,K,T,r,.04,1.5,.04,.30,-.5,paths=12000,steps=80,seed=42)
    rows.append(dict(strike=K, Black_Scholes=b, Merton=m, Heston_MC=h, Heston_MC_SE=se))
results=pd.DataFrame(rows)
results.round(3)
```

```python
fig, ax=plt.subplots(figsize=(8,4))
for col in ["Black_Scholes", "Merton", "Heston_MC"]:
    ax.plot(results.strike,results[col],marker="o",label=col)
ax.set(title="Synthetic model comparison (NOT market fit)",xlabel="Strike",ylabel="European call price")
ax.legend()
plt.tight_layout()
plt.show()
```
